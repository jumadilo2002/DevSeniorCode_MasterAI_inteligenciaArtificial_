### 1.1 — Listas y operaciones básicas

**Ejercicio 1.3** — Crea una función llamada `clasificar_esperanza_vida` que reciba un número (años) y retorne:

- `"Media"` si es entre 60 y 74.9

````python
# Tu código aquí

import pandas as pd

# Cargar el dataset

df = pd.read_csv("salud_mundial.csv")

# Dimensiones

filas, columnas = df.shape
print("Filas:", filas)
print("Columnas:", columnas)


# Tipos de datos

print("\nTipos de datos:")
print(df.dtypes)

# Valores nulos

print("\nNulos por columna:")
print(df.isnull().sum())

# Primeras 8 filas

print("\nPrimeras 8 filas:")
display(df.head(8))
````

### 2.2 — Limpieza de datos

🔧 **Ejercicio 2.2** — El dataset tiene algunos valores nulos (¡datos reales siempre tienen este problema!).  
1. Rellena los valores nulos de columnas **numéricas** con la **media** de cada columna
2. Verifica que ya no queden nulos
3. Explica en una celda de texto: ¿Por qué usamos la media y no simplemente eliminamos esas filas?

````python
# Tu código aquí

# Ver nulos antes de limpiar
print("Nulos antes de limpiar:")
print(df.isnull().sum())

# Rellenar nulos con la media en cada columna numérica
df["esperanza_vida"] = df["esperanza_vida"].fillna(df["esperanza_vida"].mean())
df["gasto_salud_usd"] = df["gasto_salud_usd"].fillna(df["gasto_salud_usd"].mean())
df["mortalidad_infantil"] = df["mortalidad_infantil"].fillna(df["mortalidad_infantil"].mean())
df["medicos_por_1000"] = df["medicos_por_1000"].fillna(df["medicos_por_1000"].mean())
df["obesidad_pct"] = df["obesidad_pct"].fillna(df["obesidad_pct"].mean())
df["diabetes_pct"] = df["diabetes_pct"].fillna(df["diabetes_pct"].mean())
df["tabaquismo_pct"] = df["tabaquismo_pct"].fillna(df["tabaquismo_pct"].mean())
df["camas_hospital_por_1000"] = df["camas_hospital_por_1000"].fillna(df["camas_hospital_por_1000"].mean())
df["vacunacion_pct"] = df["vacunacion_pct"].fillna(df["vacunacion_pct"].mean())

# Ver nulos después de limpiar
print("\nNulos después de limpiar:")
print(df.isnull().sum())
````

**Responde aquí** (doble clic para editar):

*¿Por qué usamos la media para rellenar nulos en vez de eliminar las filas?*

> 
Usamos la media porque así completamos los valores faltantes sin eliminar filas del dataset. Si borráramos esas filas, perderíamos información importante. Como son datos numéricos, la media es un valor representativo y ayuda a mantener el análisis más completo.

### 2.5 — Agrupaciones

🔧 **Ejercicio 2.5** — Usa `groupby()` para:
1. Calcular la **esperanza de vida promedio** por `region`
2. Calcular el **gasto en salud promedio** por `nivel_ingresos`
3. Calcular el **número de países** por región
4. ¿Qué región tiene la mayor esperanza de vida promedio? ¿Y la menor?

````python
# Tu código aquí

# 1. Esperanza de vida promedio por region
promedio_vida_region = df.groupby("region")["esperanza_vida"].mean()
print("Esperanza de vida promedio por region:")
print(promedio_vida_region)

# 2. Gasto en salud promedio por nivel de ingresos
promedio_gasto_ingresos = df.groupby("nivel_ingresos")["gasto_salud_usd"].mean()
print("\nGasto en salud promedio por nivel de ingresos:")
print(promedio_gasto_ingresos)

# 3. Numero de paises por region
paises_por_region = df.groupby("region")["pais"].count()
print("\nNumero de paises por region:")
print(paises_por_region)

# 4. Region con mayor y menor esperanza de vida promedio
print("\nRegion con mayor esperanza de vida promedio:")
print(promedio_vida_region.idxmax())

print("\nRegion con menor esperanza de vida promedio:")
print(promedio_vida_region.idxmin())
````

### 3.3 — Buenas prácticas

**Responde:**

**¿Qué archivos debería incluir un `.gitignore` en un proyecto de Data Science? Menciona al menos 3 y explica por qué.**

> Tu respuesta:

En un proyecto de Data Science, un .gitignore debería incluir archivos temporales, archivos del entorno virtual y archivos generados automáticamente.

Por ejemplo:

__pycache__/ porque guarda archivos temporales de Python que no es necesario subir.
venv/ o .venv/ porque contiene el entorno virtual y puede volver a crearse en otro computador.
.ipynb_checkpoints/ porque son copias automáticas que crea Jupyter Notebook y no hacen parte del trabajo final.

También podrían incluirse archivos como .DS_Store o Thumbs.db, ya que son archivos del sistema y no aportan al proyecto.

**¿Cuál es la diferencia entre `git add .` y `git add nombre_archivo`?**

> Tu respuesta:

La diferencia es que git add . agrega todos los archivos modificados de la carpeta actual al staging area, mientras que git add nombre_archivo solo agrega el archivo específico que se indique.

En otras palabras, git add . sirve para añadir todo de una vez, y git add nombre_archivo sirve cuando se quiere controlar mejor qué archivo se va a preparar para el commit.
