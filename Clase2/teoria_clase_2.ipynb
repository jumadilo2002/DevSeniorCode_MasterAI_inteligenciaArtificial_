# Diccionarios:

los diccionarios se definen utilizando llaves {} y los pares de clave-valor se separan por comas. La clave y el valor se separan por dos puntos :.

la diferencia entre una lista y un diccionario es que en una lista los elementos se acceden por su indice, mientras que en un diccionario los elementos se acceden por su clave.
